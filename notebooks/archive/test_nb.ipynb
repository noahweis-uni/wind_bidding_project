```python
import pandas as pd
from pathlib import Path

# Load day-ahead prices from standardized SMARD CSV
prices_file = Path("../data/raw/Daten zu Marktpreisen/prices_standardized.csv")
prices = pd.read_csv(prices_file, sep=";", decimal=",")
prices["timestamp"] = pd.to_datetime(prices["Datum von"], format="%d.%m.%Y %H:%M")
prices["da_price"] = prices["Preis"]

print(f"Loaded {len(prices)} price records from {prices_file.name}")
print(prices.head())
```
